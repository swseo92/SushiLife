```python
class Updater:
    def __init__(self, list_date, date):
        if not isinstance(date, pd.Timestamp):
            raise TypeError

        self.list_date = list_date
        self.date = date

        # 날짜가 업데이트 될 경우 내부 정보를 업데이트할 instance를 넣는다.
        # 해당 instance들은 "update_date" method를 갖고 있어야한다.

        self.list_instance4update = list()

        self.update()

    def set_date(self, date):
        if not isinstance(date, pd.Timestamp):
            raise TypeError

        self.date = self.list_date[np.where(self.list_date == date)][0]
        for instance in self.list_instance4update:
            instance.update_date(self.date)

    def set_instance4update(self, instance):
        self.list_instance4update.append(instance)

    def update(self):
        self.date += pd.Timedelta(days=1)

        while not (self.date == self.list_date).any():
            self.date += pd.Timedelta(days=1)

        for instance in self.list_instance4update:
            instance.update_date(instance, self.date)
```

```python
updater = Updater(list_date, pd.Timestamp(2002, 1, 2))
```

```python
for i in range(5):
    updater.update()
    print(updater.date)
```
