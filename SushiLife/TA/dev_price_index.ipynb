```python
import pandas as pd
from SushiLife import *
from SushiLife.TA import *
```

```python
f = h5py.File("../../../../data/stock_info.hdf5", "r")
array_stock, axis_stock = load_data(f, "stock", chunks=5)
array_value, axis_value = load_data(f, "value", chunks=5)

data_stock = DataAsset(array_stock, axis_stock, chunks=500)
data_value = DataAsset(array_value, axis_value, chunks=500)

updater = Updater(pd.Timestamp(2018, 12, 10), data_stock.dates)

# 거래소 생성
exchange_stock = Exchange()
exchange_stock.set_DataAsset(data_stock)

# 주식 계좌 생성
stock_account = StockAccount(exchange_stock, 출력=False)

# 거래 에이전트 생성 및 주식 계좌 등록
agent = Agent(1e8, 출력=False)
agent.set_account("stock", stock_account)

# 날짜가 변할시 업데이트 요청
updater.set_data(data_stock)
updater.set_data(data_value)

updater.set_exchange(exchange_stock)

updater.set_agent(agent)

updater.initialization()
```

```python
def TD_sequential(현재가, shift=4):
    price_flip_bear = (현재가 - np.array(pd.Series(현재가).shift(shift))) < 0
    td_seq = np.zeros_like(price_flip_bear, dtype="i")
    for i in range(len(price_flip_bear)):
        for j in range(i):
            if not price_flip_bear[i - j]:
                break
            td_seq[i] += 1
            
    return td_seq
```

```python
seq=TD_sequential(현재가)
chart.set_data(저가, mode='scatters+text', text=TD_sequential(현재가)) 
```
